**Question No 1**

```python
from collections import deque

def bfs(queue, visited, parent, other_visited, graph):

    if not queue:
        return None

    current = queue.popleft()

    print(f'Current: {current}')

    # Check if searches have already met
    if current in other_visited:
        return current

    for neighbour in graph[current]:

        if neighbour not in visited:

            visited.add(neighbour)
            parent[neighbour] = current
            queue.append(neighbour)

            # Check if searches meet at neighbour
            if neighbour in other_visited:
                return neighbour

    return None


def bidirectional_search(start, goal, graph):

    try:
        if start not in graph:
            raise KeyError

        if goal not in graph:
            raise KeyError

        if start == goal:
            print("\nAlready on Goal")
            return [start]

        forward_queue = deque([start])
        backward_queue = deque([goal])

        forward_visited = {start}
        backward_visited = {goal}

        forward_parent = {start: None}
        backward_parent = {goal: None}

        while forward_queue and backward_queue:

            # Forward Search
            meet = bfs(
                forward_queue,
                forward_visited,
                forward_parent,
                backward_visited,
                graph
            )

            if meet:

                print(f"\nMeet at : {meet}")

                path = []

                node = meet
                while node is not None:
                    path.append(node)
                    node = forward_parent[node]

                path.reverse()

                node = backward_parent[meet]
                while node is not None:
                    path.append(node)
                    node = backward_parent[node]

                return path

            # Backward Search
            meet = bfs(
                backward_queue,backward_visited,backward_parent, forward_visited,graph
            )

            if meet:

                print(f"\nMeet at : {meet}")

                path = []

                node = meet
                while node is not None:
                    path.append(node)
                    node = forward_parent[node]

                path.reverse()

                node = backward_parent[meet]
                while node is not None:
                    path.append(node)
                    node = backward_parent[node]

                return path

        print("\nNo Path Found")
        return None

    except KeyError:
        print("Given city is not in graph")
        return None


unweighted_graph = {
    city: [neighbour for neighbour, _ in neighbours]
    for city, neighbours in graph.items()
}

start = input("Enter start state: ").lower()
goal = input("Enter goal state: ").lower()



path = bidirectional_search(start, goal, unweighted_graph)

if path:
    print("\nShortest Path :", path)
```

**Question No 2**

```python
import heapq

def ucs(start, goal, graph):

    try:
        if start not in graph:
            raise KeyError

        if goal not in graph:
            raise KeyError

        if start == goal:
            print("\nAlready on goal")
            return [start], 0

        queue = []
        visited = set()

        heapq.heappush(queue, (0, {
            'current': start,
            'path': [start]
        }))

        while queue:

            cost, node = heapq.heappop(queue)

            current = node['current']
            path = node['path']

            if current in visited:
                continue

            visited.add(current)

            print(f'Current: {current}, Cost: {cost}, Path: {path}')

            if current == goal:
                print("\nGoal reached")
                return path, cost

            for neighbour, edge_cost in graph[current]:

                if neighbour not in visited:

                    heapq.heappush(queue, (cost + edge_cost,{'current': neighbour,'path': path + [neighbour]} ))

        print("\nNo path found")
        return None, None

    except KeyError:
        print("\nGiven city is not in graph")
        return None, None



start = input("Enter start state: ").lower()
goal = input("Enter goal state: ").lower()
print('\n')

path, cost = ucs(start, goal, graph)

print(f'\nShortest Path : {path}')
print(f'Total Cost    : {cost}')
```

```python

def calculate_heuristic(goal, graph):

    heuristic = {}

    for city in graph:

        priority_queue = []
        visited = set()

        heapq.heappush(priority_queue, (0, city))

        while priority_queue:

            distance, current_city = heapq.heappop(priority_queue)

            if current_city in visited:
                continue

            visited.add(current_city)

            if current_city == goal:
                heuristic[city] = distance
                break

            for next_city, road_cost in graph[current_city]:

                if next_city not in visited:

                    new_distance = distance + road_cost

                    heapq.heappush( priority_queue,(new_distance, next_city) )
    return heuristic


def best_first_search(start, goal, graph, heuristic):

    try:

        if start not in graph:
            raise KeyError

        if goal not in graph:
            raise KeyError

        if start == goal:
            print("\nAlready on Goal")
            return [start]

        priority_queue = []
        visited = set()

        heapq.heappush(priority_queue, (heuristic[start],{'current': start,'path': [start]}))

        while priority_queue:

            heuristic_value, city = heapq.heappop(priority_queue)

            current = city['current']
            path = city['path']

            if current in visited:
                continue

            visited.add(current)

            print(f'Current: {current}, Heuristic: {heuristic_value}')

            if current == goal:
                print("\nGoal Reached")
                return path

            for neighbour, road_cost in graph[current]:

                if neighbour not in visited:

                    new_path = path + [neighbour]

                    heapq.heappush(priority_queue, (heuristic[neighbour],{'current': neighbour,'path': new_path}))

        print("\nNo Path Found")
        return None

    except KeyError:
        print("\nGiven city is not in graph")
        return None


start = input("Enter start state: ").lower()
goal = input("Enter goal state: ").lower()

heuristic = calculate_heuristic(goal, graph)

print('\n')

path = best_first_search(start, goal, graph, heuristic)

print(f"\nPath : {path}")
```
