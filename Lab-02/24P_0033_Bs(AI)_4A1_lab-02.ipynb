**Graph**

```python
graph = {
    'seattle' : ['san francisco', 'denver', 'chicago'],
    'san francisco' : ['seattle', 'denver', 'los angeles'],
    'los angeles' : ['san francisco', 'denver', 'kansas city', 'dallas'],
    'denver' : ['seattle', 'san francisco', 'los angeles', 'chicago'],
    'kansas city' : ['los angeles','denver', 'chicago', 'new york', 'atlanta', 'dallas' ],
    'chicago' : ['seattle', 'denver', 'boston', 'kansas city','new york', ],
    'boston' : ['chicago', 'new york' ],
    'new york' : ['kansas city','chicago','boston','atlanta'],
    'atlanta' : ['kansas city','new york','miami','dallas','houston'],
    'miami' : ['atlanta','houston'],
    'dallas' : ['los angeles','kansas city','atlanta','houston'],
    'houston' : ['atlanta','miami','dallas']
}

```

**Question No 1**

```python
from collections import deque

def bfs(start, goal, graph):
    try:
        if start not in graph:
            raise KeyError

        if goal not in graph:
            raise KeyError

        if start == goal:
            print("\n\nAlready on goal")
            return [start]

        queue = deque()
        visited = set()

        queue.append({
            'current': start,
            'path': [start]
        })

        visited.add(start)

        while queue:

            node = queue.popleft()

            current = node['current']
            path = node['path']

            print(f'Current: {current}, Path: {path}')

            if current == goal:
                print("\n\nGoal reached")
                return path

            for neighbour in graph[current]:

                if neighbour not in visited:
                    visited.add(neighbour)

                    queue.append({
                        'current': neighbour,
                        'path': path + [neighbour]
                    })

        print("\nNo path found")
        return None

    except KeyError:
        print("\nGiven city is not in graph")
        return None



start = (input('Enter start state: ')).lower()
goal = (input('Enter goal state: ')).lower()
print('\n')
path = bfs(start, goal, graph)

print(f'shortest path : {path}')
```

```python

def dfs(start, goal, graph):
    try:
        if start not in graph:
            raise KeyError

        if goal not in graph:
            raise KeyError

        if start == goal:
            print("\n\nAlready on goal")
            return [start]

        stack = []
        visited = set()

        stack.append({
            'current': start,
            'path': [start]
        })

        visited.add(start)

        while stack:

            node = stack.pop()

            current = node['current']
            path = node['path']

            print(f'Current: {current}, Path: {path}')

            if current == goal:
                print("\n\nGoal reached")
                return path

            for neighbour in graph[current]:

                if neighbour not in visited:
                    visited.add(neighbour)

                    stack.append({
                        'current': neighbour,
                        'path': path + [neighbour]
                    })

        print("\nNo path found")
        return None

    except KeyError:
        print("\nGiven city is not in graph")
        return None



start = (input('Enter start state: ')).lower()
goal = (input('Enter goal state: ')).lower()
print('\n')
path = dfs(start, goal, graph)

print(f'DFS path : {path}')
```

**Question No 3**

```python
def iddfs(start, goal, limit, graph):

    try:
        if start not in graph:
            raise KeyError

        if goal not in graph:
            raise KeyError

        if limit <= 0:
            raise ValueError

        if start == goal:
            print("\n\nAlready on goal")
            return [start]

        for depth_limit in range(limit + 1):

            print(f'\nDepth Limit: {depth_limit}')

            stack = []

            stack.append({
                'current': start,
                'path': [start],
                'depth': 0
            })

            visited = set()
            visited.add(start)

            while stack:

                node = stack.pop()

                current = node['current']
                path = node['path']
                depth = node['depth']

                print(f'Current: {current}, Path: {path}')

                if current == goal:
                    print("\n\nGoal reached")
                    return path

                if depth < depth_limit:

                    for neighbour in graph[current]:

                        if neighbour not in visited:

                            visited.add(neighbour)

                            stack.append({
                                'current': neighbour,
                                'path': path + [neighbour],
                                'depth': depth + 1
                            })

        print("\nNo path found at this limit.")
        return None

    except KeyError:
        print("\nGiven city is not in graph")
        return None

    except ValueError:
        print("\nDepth limit must be greater than 0")
        return None


start = (input('Enter start state: ')).lower()
goal = (input('Enter goal state: ')).lower()
limit = int(input('Enter depth limit: '))

print('\n')

path = iddfs(start, goal, limit, graph)

print(f'IDDFS path : {path}')
```
