# **Question No 1**

```python
def implication(p, q):
    return (not p) or q

def biconditional(p, q):
    return p == q

def truth_table():
    try:
        values = [True, False]

        print(f'P\tQ\t~P\tP∧Q\tP∨Q\tP→Q\tP↔Q')

        for p in values:
            for q in values:
                print(
                    p, "\t",
                    q, "\t",
                    not p, "\t",
                    p and q, "\t",
                    p or q, "\t",
                    implication(p, q), "\t",
                    biconditional(p, q)
                )

    except Exception as e:
        print("Error:", e)

truth_table()
```

# By AIMA Python

```python
from logic import *

def logic_library():
    try:
        P = Symbol("P")
        Q = Symbol("Q")

        print("P AND Q")
        print(And(P, Q))

        print("P OR Q")
        print(Or(P, Q))

        print("NOT P")
        print(Not(P))

        print("P => Q")
        print(Implication(P, Q))

        print("P <=> Q")
        print(Biconditional(P, Q))

    except Exception as e:
        print("Error:", e)

logic_library()
```

# Example 1

```python
def example1():
    try:
        values = [True, False]

        print("P\tQ\tP→Q\t¬Q\tConclusion(¬P)")

        for p in values:
            for q in values:

                premise1 = (not p) or q      # P → Q
                premise2 = not q             # ¬Q

                if premise1 and premise2:
                    print(p, "\t", q, "\t", premise1, "\t", premise2, "\t", not p)

    except Exception as e:
        print("Error:", e)

example1()
```

# Example 2


```python
def example2():
    try:
        values = [True, False]

        print("P\tQ\tP∨Q\t¬P\tConclusion(Q)")

        for p in values:
            for q in values:

                premise1 = p or q
                premise2 = not p

                if premise1 and premise2:
                    print(p, "\t", q, "\t", premise1, "\t", premise2, "\t", q)

    except Exception as e:
        print("Error:", e)

example2()
```

# Example 3

```python
def example3():
    try:
        values = [True, False]

        print("P\tQ\tConclusion(P∨Q)")

        for p in values:
            for q in values:

                if p and q:
                    print(p, "\t", q, "\t", p or q)

    except Exception as e:
        print("Error:", e)

example3()
```
