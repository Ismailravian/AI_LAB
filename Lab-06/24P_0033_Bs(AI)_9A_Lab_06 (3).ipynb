```python
graph = {
    'wa' : ['nt', 'sa'],
    'nt' : ['wa', 'sa', 'q'],
    'sa' : ['wa', 'nt', 'q', 'nsw', 'v'],
    'q' : ['nt', 'sa', 'nsw'],
    'nsw' : ['q', 'sa', 'v'],
    'v' : ['sa', 'v'],
    't' : []
}
print(graph)

colors = ["Red", "Green", "Blue"]

domains = {}

for state in graph:
    domains[state] = colors.copy()
```

# **Question No 1**

```python
from collections import deque

def ac3(graph, domains):

    queue = deque()

    for node in graph:
        for neighbour in graph[node]:
            queue.append((node, neighbour))
    print(queue)
    while queue:

        xi, xj = queue.popleft()

        revised = False

        for color in domains[xi][:]:

            found = False

            for c in domains[xj]:
                if color != c:
                    found = True
                    break

            if not found:
                domains[xi].remove(color)
                revised = True

        if revised:

            if len(domains[xi]) == 0:
                return False

            for neighbour in graph[xi]:
                if neighbour != xj:
                    queue.append((neighbour, xi))

    return True

ac3(graph, domains)
print('\n\nDomain :')
for state in domains:
  print(state, ":", domains[state])
```

# **Question No 2**

```python
def backtracking(assignment):

    if len(assignment) == len(graph):
        return assignment

    for state in graph:

        if state not in assignment:

            for color in domains[state]:

                safe = True

                for neighbour in graph[state]:

                    if neighbour in assignment:

                        if assignment[neighbour] == color:
                            safe = False
                            break

                if safe:

                    assignment[state] = color

                    answer = backtracking(assignment)

                    if answer:
                        return answer

                    del assignment[state]

            return None


solution = backtracking({})
for state in solution:
  print(state, ":", solution[state])
```
