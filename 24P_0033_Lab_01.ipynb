# **Question No 1**

```python

def calculate_grade(marks):
    if marks >= 75 :
        print(f"Grade is A" )
    elif marks >= 60 and marks < 75 :
        print(f"Grade is B" )
    elif marks >= 40  and marks < 60 :
        print(f"Grade is C" )
    else :
        print(f"Grade is Fail" )


no_students = int(input("Enter No of students: "))
no_subjects = int(input("Enter no of subjects: "))

fail_stu = []
average = 0

for i in range(no_students):
    name = input("\n\nEnter name of student: ")
    total = 0


    for j in range(no_subjects):
        print(f'Enter maks of subject {j+1}', end = ' ')
        total = total + int(input())


    print(f'Name : {name}')
    average = total / no_subjects
    print(f'Average: {average}')
    calculate_grade(average)
    if average < 40 :
        fail_stu.append(name)


print('\n\nFail students or student at risk')
if len(fail_stu) == 0 :
    print('No fail student')
else :
    for i in range(len(fail_stu)):
        print(fail_stu[i])

```

# **Question No 2**

```python


transactions = int(input("Enter number of sales transactions: "))

total_revenue = 0
product_quantity = {}



for i in range(transactions):

    print(f"\nTransaction {i+1}")

    product_id = int(input("Enter Product ID: "))
    quantity = int(input(" enter Quantity Sold: "))
    price = float(input("Enter Price per Unit: "))

    total_revenue = total_revenue + (quantity * price)

    if product_id in product_quantity:
        product_quantity[product_id] = product_quantity[product_id] + quantity
    else:
        product_quantity[product_id] = quantity



highest_product = None
highest_quantity = 0

for product_id in product_quantity:
    if product_quantity[product_id] > highest_quantity:
        highest_quantity = product_quantity[product_id]
        highest_product = product_id


print("\n\nTotal Revenue:", total_revenue)

print(' Highest Quantity Sold:', highest_product)
print("Total quantity Sold:", highest_quantity)

print("\n\nProducts with Zero Quantity:")



found = False

for product_id in product_quantity:
    if product_quantity[product_id] == 0:
        print(product_id)
        found = True

if found == False:
    print("None")
```

# **Question No 5**

```python
city_tracker = {}

num_cities = int(input("Enter number of cities: "))





for i in range(num_cities):
    city = input("\nEnter city name: ")

    num_connections = int(input(f"No of cities are directly connected to {city} "))

    connections = []

    for j in range(num_connections):
        connected_city = input(f"Enter connected city {j+1}: ")
        connections.append(connected_city)

    city_tracker[city] = connections




print("\ncities and Their Connections")
for city in city_tracker:
    print(city, "->", city_tracker[city])


city1 = input('\nEnter departure city: ')
city2 = input("Enter destination city: ")



if city1 in city_tracker and city2 in city_tracker[city1]:
    print(f"Direct connection exists between {city1} and {city2}")
else:
    print(f"No direct connection between {city1} and {city2}")


max_city = ' '
max_connections = 0

for city in city_tracker:
    if len(city_tracker[city]) > max_connections:
        max_connections = len(city_tracker[city])
        max_city = city

print('\n\nCity with Most Connections:')
print(max_city, "with", max_connections, 'connections')
```
