# **Question No 1**

```python
import heapq

def a_steric(start, goal, graph, hurestic):

    try:
        if start not in graph:
            raise KeyError

        if start == goal:
            print("\nAlready on goal")
            return [start], 0

        queue = []
        visited = set()

        heapq.heappush(queue, (0, {
            'current': start,
            'path': [start]
        }))

        while queue:

            cost, node = heapq.heappop(queue)

            current = node['current']
            path = node['path']

            if current in visited:
                continue

            visited.add(current)

            print(f'Current: {current}, Cost: {cost}, Path: {path}')

            if current == goal:
                print("\nGoal reached")
                return path, cost

            for neighbour, edge_cost in graph[current]:

                if neighbour not in visited:

                    actual_cost = cost + edge_cost
                    h_cost = actual_cost + hurestic[current]

                    heapq.heappush(queue, ( actual_cost,{
                        'current': neighbour,
                        'path': path + [neighbour]
                    } ))

        print("\nNo path found")
        return None, None

    except KeyError:
        print("\nGiven city is not in graph")
        return None, None



start = input("Enter start state: ").lower()

goal = 'bucharest'
print('\n')

path, cost = a_steric(start, goal, graph, hurestic)

print(f'\nShortest Path : {path}')
print(f'Total Cost    : {cost}')
```

# **Question No 2(with Iterrative approach)**

```python
def minimum(leaf_values):
    new_leaf = []
    for i in range(0, len(leaf_values), 2):
        new_leaf.append(min(leaf_values[i], leaf_values[i + 1]))
    return new_leaf


def maximum(leaf_values):
    new_leaf = []
    for i in range(0, len(leaf_values), 2):
        new_leaf.append(max(leaf_values[i], leaf_values[i + 1]))
    return new_leaf


def minimax(leaf_values, depth):
    while depth > 0:
        if depth == 1:
            return leaf_values[0]

        if depth % 2 == 0:
            leaf_values = minimum(leaf_values)
        else:
            leaf_values = maximum(leaf_values)

        depth -= 1

        return leaf_values[0]


leaf_values = [3, 5, 6, 9, 1, 2, 0, -1]
depth = 3

value = minimax(leaf_values, depth)
print(value)
```

# **Question No 2 (Recursive approach)**

```python
def isterminal(leaf_values):
    return isinstance(leaf_values, int)


def minimum(leaf_values):

    a = minimax(leaf_values[0], True)
    b = minimax(leaf_values[1], True)

    return min(a, b)


def maximum(leaf_values):

    a = minimax(leaf_values[0], False)
    b = minimax(leaf_values[1], False)

    return max(a, b)


def minimax(leaf_values, is_max=True):

    if isterminal(leaf_values):
        return leaf_values

    if is_max:
        return maximum(leaf_values)
    else:
        return minimum(leaf_values)


leaf_values = [[[3, 5], [6, 9]], [[1, 2], [0, -1]]]

value = minimax(leaf_values)

print(value)
```

# **Question No 3**

```python
def isterminal(leaf_values):
    return isinstance(leaf_values, int)


def minimum(leaf_values, alpha, beta):

    value = float('inf')

    for child in leaf_values:

        value = min(value, alphabeta(child, True, alpha, beta))
        beta = min(beta, value)

        if beta <= alpha:
            break

    return value


def maximum(leaf_values, alpha, beta):

    value = float('-inf')

    for child in leaf_values:

        value = max(value, alphabeta(child, False, alpha, beta))
        alpha = max(alpha, value)

        if beta <= alpha:
            break

    return value


def alphabeta(leaf_values, is_max=True, alpha=float('-inf'), beta=float('inf')):

    if isterminal(leaf_values):
        return leaf_values

    if is_max:
        return maximum(leaf_values, alpha, beta)
    else:
        return minimum(leaf_values, alpha, beta)


leaf_values = [[[3, 5], [6, 9]], [[1, 2], [0, -1]]]

value = alphabeta(leaf_values)

print(value)
```
